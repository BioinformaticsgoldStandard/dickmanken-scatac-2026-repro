# Setup: download data and notebooks

Run this notebook first, before any of the PUMATAC analysis notebooks (numbered `0_...` to `5_...` inside `notebooks_PUMATAC/`).

```python
"""
Download data and notebooks needed for the reproducibility analysis.
Run this notebook first, before any of the numbered analysis notebooks.

Note on data provenance: the mouse cortex 10x Genomics sample used here
(8k_mouse_cortex_ATACv2_nextgem_Chromium_Controller) is downloaded directly
from the 10x Genomics website, as referenced in the article's Data
Availability section - not from the GEO accession (GSE293575), which
covers the HyDrop v2-generated data instead.
"""
import os
import subprocess
import sys

# Anchor paths to this notebook's own location rather than assuming a
# fixed working directory, so this still works if Jupyter's cwd differs
# from the notebook's folder.
NOTEBOOK_DIR = os.path.dirname(os.path.abspath("__file__"))
REPO_ROOT = os.path.abspath(os.path.join(NOTEBOOK_DIR, ".."))
SCRIPTS_DIR = os.path.join(REPO_ROOT, "scripts")

if REPO_ROOT not in sys.path:
    sys.path.append(REPO_ROOT)

from scripts.download_data import download_and_verify


def run_script(script_name):
    """Run a script as a subprocess using the current Python interpreter,
    and stop immediately (raising an error) if it fails. This mirrors the
    fail-fast behaviour already used in download_data.py, so a partial
    failure here is never silently ignored."""
    script_path = os.path.join(SCRIPTS_DIR, script_name)
    result = subprocess.run([sys.executable, script_path])
    if result.returncode != 0:
        raise RuntimeError(
            f"{script_name} exited with code {result.returncode}. "
            "Stopping setup - fix the error above before re-running this notebook."
        )


print("=" * 60)
print("Step 1: Downloading notebooks from PUMATAC tutorial")
print("=" * 60)
run_script("download_notebooks.py")

print("\n" + "=" * 60)
print("Step 2: Applying known patches to downloaded notebooks")
print("=" * 60)
# The notebooks downloaded in Step 1 come straight from the upstream
# PUMATAC_tutorial repository and are overwritten on every run of this
# setup notebook, so any fix has to be applied here, in code, rather than
# edited by hand directly in those files (see scripts/patch_notebooks.py
# for the full list of patches and the reasoning behind each one).
run_script("patch_notebooks.py")

print("\n" + "=" * 60)
print("Step 3: Downloading and verifying 10x Genomics data")
print("=" * 60)
# Downloads (if missing or corrupted), verifies the MD5 checksum, and
# extracts the FASTQ files. Raises SystemExit if the checksum does not match.
download_and_verify()

print("\n" + "=" * 60)
print("Setup complete! You can now run the numbered analysis notebooks.")
print("=" * 60)
```
